```python
#Load in libraries and dataset
import pandas as pd

df = pd.read_csv("Nashville_housing_data.csv")

#Print the first 5 rows
df.head()
```

```python
#Profile the data
df.info()
df.describe()
```

```python
#Standardize headings
df = df.rename(columns={'Unnamed: 0.1':'unnamed_0.1', 
                       'Unnamed: 0':'unnamed_0',
                       'Parcel ID': 'parcel_id', 
                       'Land Use': 'land_use', 
                       'Property Address':'property_address', 
                       'Suite/Condo':'suite_condo',
                       'Property City':'property_city',
                       'Sale Date': 'sale_date', 
                       'Sale Price': 'sale_price', 
                       'Legal Reference':'legal_reference',
                       'Sold As Vacant':'sold_as_vacant',
                       'Multiple Parcels Involved in Sale':'multiple_parcels_involved_in_sale',
                       'Owner Name':'owner_name', 
                       'Address':'address',
                       'City':'city',
                       'State':'state',
                       'Acreage':'acreage',
                       'Tax District':'tax_district',
                       'Neighborhood':'neighborhood',
                       'Land Value':'land_value', 
                       'Building Value':'building_value',
                       'Total Value':'total_value', 
                       'Finished Area': 'finished_area', 
                       'Foundation Type':'foundation_type', 
                       'Year Built':'year_built',
                       'Exterior Wall':'exterior_wall', 
                       'Grade':'grade',
                       'Bedrooms':'bedrooms',
                       'Full Bath':'full_bath', 
                       'Half Bath':'half_bath'
                      })
df.head()
```

```python
#Eliminate unneeded columns
df = df.drop(columns=["unnamed_0.1", "unnamed_0"], errors="ignore")
df.head()
```

```python
# Handle missing values
df = df.dropna(subset=['sale_price'])
df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].median())

df.head()
```

```python
#Fix column data types
df['sale_date'] = pd.to_datetime(df['sale_date'], errors='coerce')
df.head()
```

```python
#Clean inconsistent categorical data
df['sold_as_vacant'] = df['sold_as_vacant'].str.strip().str.capitalize()
df.head()
```

```python
#Save the cleaned dataset
df.to_csv("nashville_housing_cleaned.csv", index=False)
```
